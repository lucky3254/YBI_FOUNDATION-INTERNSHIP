# Python Dictionary

```python
# =====================================================================
# WHAT IS A PYTHON DICTIONARY?
# =====================================================================
# In simple words, a Python dictionary is like a real-world dictionary
# or a phonebook.
#
# - In a real dictionary, you look up a "word" to find its "definition".
# - In a phonebook, you look up a "name" to find their "phone number".
#
# In Python, we call the "word" or "name" a KEY, and the "definition"
# or "phone number" a VALUE.
# Together, they form a KEY-VALUE pair.
#
# Dictionaries are written with curly brackets {} and look like this:
# my_dictionary = { "key": "value" }
#
# Key Rules:
# 1. Keys must be unique (you can't have two identical keys).
# 2. Keys are used to quickly find and retrieve the values.

print("--- Python Dictionary Guide with 5 Simple Examples ---\n")

# =====================================================================
# EXAMPLE 1: The Phonebook (Connecting Names to Numbers)
# =====================================================================
# This is the classic example. We want to store contact numbers.
# Key: Person's Name (string)
# Value: Phone Number (string or integer)

phonebook = {
    "Alice": "555-0199",
    "Bob": "555-0142",
    "Charlie": "555-0187"
}

print("Example 1: Phonebook Dictionary")
print(phonebook)
# How to access a specific value using its key:
alices_number = phonebook["Alice"]
print(f"Alice's phone number is: {alices_number}\n")


# =====================================================================
# EXAMPLE 2: A Toy Box (Connecting Toy Names to Their Colors)
# =====================================================================
# Let's say we have different toys and we want to remember what color each toy is.
# Key: Toy Name
# Value: Color

toy_colors = {
    "ball": "red",
    "toy car": "blue",
    "teddy bear": "brown",
    "blocks": "yellow"
}

print("Example 2: Toy Colors Dictionary")
print(toy_colors)
# Accessing the color of the toy car:
car_color = toy_colors["toy car"]
print(f"The toy car is: {car_color}\n")


# =====================================================================
# EXAMPLE 3: Student Grades (Connecting Names to Scores)
# =====================================================================
# A teacher wants to keep track of exam scores for students.
# Key: Student Name
# Value: Grade (Percentage/Number)

student_grades = {
    "Emma": 95,
    "Liam": 88,
    "Olivia": 92,
    "Noah": 79
}

print("Example 3: Student Grades")
print(student_grades)
# Let's check how Liam did:
liams_grade = student_grades["Liam"]
print(f"Liam scored: {liams_grade}%\n")


# =====================================================================
# EXAMPLE 4: Recipe Ingredients (Connecting Item Names to Quantities)
# =====================================================================
# Dictionaries are great for shopping lists or recipes where we need amounts.
# Key: Ingredient Name
# Value: How much we need (string representing amount)

cake_recipe = {
    "Flour": "2 cups",
    "Sugar": "1 cup",
    "Eggs": "3 large",
    "Butter": "1/2 cup"
}

print("Example 4: Cake Recipe Ingredients")
for ingredient, amount in cake_recipe.items():
    print(f"- {ingredient}: {amount}")
print() # New line


# =====================================================================
# EXAMPLE 5: A Video Game Character (Multi-type Data Dictionary)
# =====================================================================
# Dictionaries can store different types of information about one single object.
# This is incredibly useful in programming.
# Keys: Attributes (strings)
# Values: Mixed types (strings, integers, lists)

game_character = {
    "name": "Shadow Knight",
    "level": 42,
    "health": 150,
    "is_alive": True,
    "inventory": ["sword", "shield", "health potion"]
}

print("Example 5: Video Game Character Attributes")
print(f"Character Name: {game_character['name']}")
print(f"Level: {game_character['level']}")
print(f"Items in inventory: {game_character['inventory']}")
```
