# Pandas Data Summary

```python
# Pandas Data Summary Example
# Common methods to summarize a DataFrame include:
# 1. df.info()       - Overview of data types, non-null counts, and memory usage.
# 2. df.describe()   - Summary statistics (mean, std, min, max, percentiles) for numerical columns.
# 3. df.value_counts() - Frequency counts of unique values in a specific column.

import pandas as pd

# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 35, 40, 25, None],
    'Salary': [50000, 60000, 70000, 80000, 52000, 65000],
    'Department': ['HR', 'IT', 'IT', 'HR', 'Marketing', 'IT']
}
df = pd.DataFrame(data)

print("--- 1. Dataset Information (df.info) ---")
df.info()

print("\n--- 2. Statistical Summary (df.describe) ---")
print(df.describe())

print("\n--- 3. Department Value Counts (df['Department'].value_counts) ---")
print(df['Department'].value_counts())
```

### Comprehensive Guide to Summarizing Data in Pandas

When working with datasets in Pandas, summarizing your data is a crucial step in exploratory data analysis (EDA). It allows you to understand the structure, distributions, missing values, and key statistics of your dataset. Below is a detailed breakdown of the most essential Pandas methods used to summarize data.

---

#### 1. High-Level Dataset Metadata: `df.info()`
The `info()` method provides a concise summary of your DataFrame, focusing on its architecture and resource usage rather than statistics.
* **What it tells you:**
  * **Total Range Index:** The number of rows (e.g., `RangeIndex: 6 entries, 0 to 5`).
  * **Data Columns:** The total number of columns.
  * **Column Inventory:** A table showing the name of each column, the count of non-null (non-missing) values, and its specific data type (`Dtype`).
  * **Memory Footprint:** The approximate RAM used by the DataFrame.
* **Why it matters:** It is the fastest way to detect missing data (by comparing non-null counts against total rows) and verify if data types were loaded correctly (e.g., dates loaded as text `object` types instead of `datetime`).

---

#### 2. Descriptive Statistics: `df.describe()`
The `describe()` method calculates central tendency, dispersion, and shape of a dataset’s distribution. By default, it processes only numeric columns.
* **Default Numeric Metrics:**
  * `count`: The number of valid, non-null data points.
  * `mean`: The arithmetic average.
  * `std`: The standard deviation (measures dispersion relative to the mean).
  * `min` & `max`: Boundary values of the dataset.
  * `25%`, `50%` (Median), `75%`: Quartiles that show the distribution profile.
* **For Categorical Data (with `include='object'` or `include='all'`):**
  * `unique`: The count of distinct values.
  * `top`: The most frequent value (the mode).
  * `freq`: The frequency of the top value.

---

#### 3. Counting Categorical Values: `df['col'].value_counts()`
This method operates on a single column (Series) to count occurrences of unique elements.
* **Key Parameters:**
  * `normalize=True`: Converts absolute counts into relative frequencies (percentages).
  * `dropna=False`: Includes missing (`NaN`) values in the counts to help track incomplete categorical entries.

#### 4. Additional Essential Summary Functions
Apart from the main three, Pandas provides several granular tools for data summary:
* **`df.shape`**: A simple tuple representing (number of rows, number of columns).
* **`df.dtypes`**: Returns only the data types of each column.
* **`df.isna().sum()`**: Counts the number of missing values in each column.
* **`df.nunique()`**: Counts unique values for each column across the whole DataFrame.

```python
# Detailed Example Demonstration
import pandas as pd
import numpy as np

# Create an expanded dataset for demonstration
rich_data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    'Age': [25, 30, 35, 40, 25, np.nan, 28, 52],
    'Salary': [50000, 60000, 70000, 80000, 52000, 65000, np.nan, 120000],
    'Department': ['HR', 'IT', 'IT', 'HR', 'Marketing', 'IT', 'Marketing', 'Executive'],
    'Remote': [True, False, True, False, True, True, False, False]
}

rich_df = pd.DataFrame(rich_data)

print("=== SHAPE OF THE DATAFRAME ===")
print(rich_df.shape)

print("\n=== MISSING VALUES PER COLUMN ===")
print(rich_df.isna().sum())

print("\n=== DESCRIBING CATEGORICAL COLUMNS ===")
print(rich_df.describe(include='object'))

print("\n=== DEPARTMENT PROPORTIONS (NORMALIZED VALUE COUNTS) ===")
print(rich_df['Department'].value_counts(normalize=True) * 100)
```
