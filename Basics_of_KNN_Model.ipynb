# Basics of KNN Model

```markdown
## What is K-Nearest Neighbors (KNN)?

**K-Nearest Neighbors (KNN)** is a simple, non-parametric, and lazy learning algorithm used for both **classification** and **regression** tasks.

### How KNN Works Step-by-Step:
1. **Choose the value of $K$**: Select the number of nearest neighbors you want to vote on the class of a new data point.
2. **Calculate Distance**: Measure the distance between the query point and all other training data points. Common distance metrics include:
   - **Euclidean Distance** (straight-line distance):
     $$d(p, q) = \sqrt{\sum_{i=1}^{n} (p_i - q_i)^2}$$
   - **Manhattan Distance** (grid-like path distance):
     $$d(p, q) = \sum_{i=1}^{n} |p_i - q_i|$$
3. **Find the Nearest Neighbors**: Sort the distances and identify the top $K$ closest training data points.
4. **Vote**:
   - For **Classification**: Assign the class label that is most common among its $K$ nearest neighbors.
   - For **Regression**: Predict the value based on the average (or weighted average) of the values of the $K$ nearest neighbors.

### Advantages of KNN:
- **Simple to understand** and implement.
- No training phase required (making it a "lazy learner").
- Naturally handles multi-class classification.

### Disadvantages of KNN:
- **Computationally expensive** during testing because it calculates the distance to every single point in the dataset.
- **Sensitive to noisy data**, outliers, and the scale of the features (which is why feature scaling is critical).
```

```markdown
In this notebook, we will implement a standard **K-Nearest Neighbors (KNN)** classifier using the `scikit-learn` library with the classic Iris dataset.
```

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score
import pandas as pd

# 1. Load data
iris = load_iris()
X, y = iris.data, iris.target

# Preview as a pandas DataFrame
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y
display(df.head())
````

```markdown
### Feature Scaling and Training
KNN is sensitive to dataset scale, so we use `StandardScaler` to normalize the dimensions before fitting the model.
```

````python
# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 3. Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Initialize and fit the KNN model (K=3)
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled, y_train)

# 5. Evaluate the model
y_pred = knn.predict(X_test_scaled)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}\n")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
````

```markdown
### Finding the Optimal Value of K using Cross-Validation
To choose the best $K$, we can evaluate our model using $k$-fold cross-validation on the training set for different values of $K$. This helps us avoid overfitting to our specific split.
```

````python
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt

# Try K values from 1 to 20
k_values = list(range(1, 21))
cv_scores = []

# Perform 10-fold cross-validation for each K
for k in k_values:
    knn_cv = KNeighborsClassifier(n_neighbors=k)
    # We use the scaled training data
    scores = cross_val_score(knn_cv, X_train_scaled, y_train, cv=10, scoring='accuracy')
    cv_scores.append(scores.mean())

# Find the optimal K (the index with the maximum accuracy score)
optimal_k = k_values[cv_scores.index(max(cv_scores))]
print(f"The optimal number of neighbors is K = {optimal_k} with a cross-validation score of {max(cv_scores):.4f}\n")

# Plot the cross-validation accuracies
plt.figure(figsize=(10, 6))
plt.plot(k_values, cv_scores, marker='o', linestyle='dashed', color='blue', markerfacecolor='red', markersize=8)
plt.title('Accuracy vs. K Value (10-Fold Cross-Validation)')
plt.xlabel('Value of K for KNN')
plt.ylabel('Cross-Validated Accuracy')
plt.xticks(k_values)
plt.grid(True)
plt.show()
````

````python
# Real-world Example: Simple Movie Recommendation System using KNN
import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors

# 1. Create a dummy movie dataset
# Features: [Action Intensity (0-10), Comedy Level (0-10)]
movie_data = {
    'Movie Name': [
        'Die Hard', 'Lethal Weapon', 'Superbad', 'The Hangover',
        'Mad Max', 'Step Brothers', 'John Wick', 'Dumb and Dumber'
    ],
    'Action': [9, 8, 1, 2, 10, 1, 10, 2],
    'Comedy': [2, 3, 9, 8, 1, 9, 1, 10]
}

df_movies = pd.DataFrame(movie_data)
X_movies = df_movies[['Action', 'Comedy']].values

# 2. Initialize unsupervised NearestNeighbors model
# We use 2 neighbors (excluding the movie itself)
knn_recommender = NearestNeighbors(n_neighbors=3, metric='euclidean')
knn_recommender.fit(X_movies)

# 3. Recommend similar movies for "John Wick" (index 6 in our list)
query_index = 6
query_features = X_movies[query_index].reshape(1, -1)

distances, indices = knn_recommender.kneighbors(query_features)

print(f"If you liked '{df_movies.iloc[query_index]['Movie Name']}', you might also like:")
for i in range(1, len(indices[0])):
    recommended_idx = indices[0][i]
    print(f"- {df_movies.iloc[recommended_idx]['Movie Name']} (Distance: {distances[0][i]:.2f})")
````
