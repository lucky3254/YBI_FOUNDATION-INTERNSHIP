# Python Loops

```python
# Example of a 'for' loop iterating over a range
print("For loop with range:")
for i in range(5):
    print(f"Iteration {i}")

print("\nFor loop iterating over a list:")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# Example of a 'while' loop
print("\nWhile loop example:")
count = 0
while count < 3:
    print(f"Count is {count}")
    count += 1
```

### Understanding Python Loops

Loops are fundamental programming constructs used to repeat a block of code multiple times. Python provides two primary types of loops:

1. **`for` Loops**:
   - Used for **definite iteration** (when you know beforehand how many times the loop should run).
   - Iterates over any sequence (such as a list, tuple, dictionary, set, or string) or an iterable object like `range()`.

2. **`while` Loops**:
   - Used for **indefinite iteration** (when the loop needs to run until a specific condition is no longer met).
   - Must always have a mechanism to eventually make the condition `False` to prevent an infinite loop.

### Loop Control Statements
- **`break`**: Terminates the loop prematurely.
- **`continue`**: Skips the rest of the current iteration and jumps directly to the next iteration.

```python
# 1. Demonstrating a 'for' loop with loop control (break)
print("--- For Loop with Break ---")
for number in range(1, 10):
    if number == 5:
        print("Found 5! Breaking the loop.")
        break
    print(f"Number: {number}")

# 2. Demonstrating a 'while' loop with loop control (continue)
print("\n--- While Loop with Continue ---")
x = 0
while x < 5:
    x += 1
    if x == 3:
        print("Skipping 3...")
        continue
    print(f"x is {x}")
```

**Question 1:** Write a loop that calculates the sum of all numbers from `1` to `100`.

**Question 2:** Given the list `numbers = [12, 16, 21, 30, 45, 50]`, write a `for` loop that iterates through the list and only prints numbers that are divisible by `5`. Use a `continue` statement to skip the others.

**Question 3:** Write a `while` loop that asks the user for input and repeats until they type "exit".

```python
# === Answer 1: Summing numbers 1 to 100 ===
print("--- Answer 1 ---")
total_sum = 0
for number in range(1, 101):  # 101 is exclusive, so it stops at 100
    total_sum += number
print(f"The sum of numbers from 1 to 100 is: {total_sum}")


# === Answer 2: Filtering numbers divisible by 5 ===
print("\n--- Answer 2 ---")
numbers = [12, 16, 21, 30, 45, 50]
print("Numbers divisible by 5:")
for num in numbers:
    if num % 5 != 0:
        continue  # Skip this number if it is not divisible by 5
    print(num)


# === Answer 3: While loop with exit condition ===
print("\n--- Answer 3 (Mock Run) ---")
# Since we are in a non-interactive output block, we mock the user typing "exit" on their third try:
mock_inputs = ["hello", "python", "exit"]
input_index = 0

while True:
    user_input = mock_inputs[input_index]
    print(f"User typed: {user_input}")
    if user_input == "exit":
        print("Loop exited successfully!")
        break
    input_index += 1
```
