# Python Tuples

```python
# What is a Tuple in Python?
# A tuple is a built-in, ordered, and immutable collection of elements.
# Once a tuple is created, you cannot change, add, or remove its items.

# How Tuples Work (Their Purpose):
# 1. Data Integrity: Because they are immutable, they protect data from accidental modification.
# 2. Performance: Tuples are slightly faster than lists, making them efficient for read-only data.
# 3. Dictionary Keys: Since they are immutable, they can be used as keys in dictionaries (unlike lists).

# --- Example of a Tuple ---

# Creating a tuple of coordinates (latitude, longitude)
coordinates = (40.7128, -74.0060)

# Accessing elements by index
latitude = coordinates[0]
longitude = coordinates[1]

print("Tuple:", coordinates)
print(f"Latitude: {latitude}, Longitude: {longitude}")

# Trying to modify a tuple will raise a TypeError
try:
    coordinates[0] = 34.0522
except TypeError as e:
    print("\nError when trying to modify the tuple:", e)

# Tuples can also store mixed data types and support unpacking
user_info = ("Alice", 30, "Engineer")
name, age, profession = user_info
print(f"\nUnpacked Info - Name: {name}, Age: {age}, Profession: {profession}")
```

### Differences Between Tuples and Lists in Python

While both tuples and lists are sequence data types that can store collections of items, they have fundamental differences:

| Feature | List | Tuple |
| :--- | :--- | :--- |
| **Mutability** | **Mutable** (can be changed after creation) | **Immutable** (cannot be changed after creation) |
| **Syntax** | Defined using square brackets `[]` | Defined using parentheses `()` |
| **Size/Memory** | Allocates extra memory to support dynamic changes | Allocates only the exact memory required (more memory-efficient) |
| **Performance** | Slower than tuples | Faster execution and iteration |
| **Methods** | Supports many built-in methods like `.append()`, `.remove()`, `.pop()` | Supports only `.count()` and `.index()` |
| **Use Case** | Used for collections of homogeneous data that change | Used for heterogeneous or read-only data (e.g., coordinates, DB records) |

```python
import sys

# 1. Mutability Demonstration
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

my_list[0] = 100
print("Modified List:", my_list)

try:
    my_tuple[0] = 100
except TypeError as e:
    print("Tuple Mutation Error:", e)

# 2. Memory Size Comparison
list_memory = sys.getsizeof([1, 2, 3, 4, 5])
tuple_memory = sys.getsizeof((1, 2, 3, 4, 5))

print(f"\nMemory occupied by list: {list_memory} bytes")
print(f"Memory occupied by tuple: {tuple_memory} bytes")
```

### When is it Better to Use a List Instead of a Tuple?

You should choose a **List** over a Tuple in the following scenarios:

1. **Dynamic Collections (Adding/Removing Items):** If you need to append, insert, remove, or modify items as the program runs.
2. **Homogeneous Data with Frequent Changes:** When all items represent the same kind of entity (e.g., a list of usernames, temperatures, or prices) that frequently changes.
3. **In-place Operations (Sorting/Shuffling/Reversing):** Lists allow you to sort or reverse items without creating a new copy in memory.
4. **Implementing Data Structures:** Lists are essential when you need to model Stacks (LIFO) or Queues (FIFO) using operations like `.append()` and `.pop()`.

```python
# 1. Dynamic Growth: Lists can grow or shrink dynamically
shopping_list = []
shopping_list.append("apples")
shopping_list.append("bananas")
print("Current Shopping List:", shopping_list)

# 2. In-place Sorting: We can sort a list directly without creating a new variable
random_numbers = [5, 2, 9, 1]
random_numbers.sort()  # Sorts the list in place
print("Sorted List (In-place):", random_numbers)

# 3. Stack Operations (LIFO): Using a list to push and pop items
stack = []
stack.append("Page 1")  # Push
stack.append("Page 2")
print("\nBrowsing History Stack:", stack)
last_visited = stack.pop()  # Pop
print("Popped item:", last_visited)
print("Remaining Stack:", stack)
```
