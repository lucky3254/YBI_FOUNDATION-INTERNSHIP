# Basics of Linear Regression Model

### What is Linear Regression?
Linear Regression is a fundamental statistical and machine learning technique used to model the relationship between a dependent variable (target) and one or more independent variables (features).

In simple linear regression (one independent variable), the relationship is modeled using a straight line:

$$y = \beta_0 + \beta_1 x + \epsilon$$

- **$y$**: Dependent variable / target
- **$x$**: Independent variable / feature
- **$\beta_0$**: Y-intercept (the value of $y$ when $x=0$)
- **$\beta_1$**: Slope of the line (how much $y$ changes for a unit change in $x$)
- **$\epsilon$**: Random error term

Let's demonstrate this with a simple Python simulation. We will generate some synthetic linear data, add random noise, and then plot the relationship.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data
np.random.seed(42)
x = 2 * np.random.rand(100, 1)
y = 4 + 3 * x + np.random.randn(100, 1)

# Plot the data
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Data points')
plt.xlabel('Independent Variable (x)')
plt.ylabel('Dependent Variable (y)')
plt.title('Relationship between x and y')
plt.legend()
plt.grid(True)
plt.show()
```

### Fitting the Model with Scikit-Learn
Now, we will use the `LinearRegression` class from `scikit-learn` to find the best-fitting line (estimating $\beta_0$ and $\beta_1$) for our synthetic dataset.

```python
from sklearn.linear_model import LinearRegression

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model to our data
model.fit(x, y)

# Retrieve the intercept (beta_0) and slope (beta_1)
beta_0 = model.intercept_[0]
beta_1 = model.coef_[0][0]

print(f"Estimated Intercept (beta_0): {beta_0:.4f} (True value: 4.0)")
print(f"Estimated Slope (beta_1): {beta_1:.4f} (True value: 3.0)")
```

Let's plot the estimated regression line over our original scatter plot of data points.

```python
# Predict y values using the fitted model
y_pred = model.predict(x)

# Plot the actual data points and the regression line
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Data points')
plt.plot(x, y_pred, color='red', linewidth=2, label='Fitted Regression Line')
plt.xlabel('Independent Variable (x)')
plt.ylabel('Dependent Variable (y)')
plt.title('Linear Regression Model Fit')
plt.legend()
plt.grid(True)
plt.show()
```
