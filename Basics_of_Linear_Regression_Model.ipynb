# Basics of Linear Regression Model

## 1. What is Linear Regression?

**Linear Regression** is a supervised machine learning algorithm used to find the relationship between an **independent variable (X)** and a **dependent variable (Y)**.

It is mainly used for **predicting a continuous numerical value**.

### Simple Example

Suppose we want to predict a student's marks based on the number of hours they study.

- **X = Study Hours**
- **Y = Marks**

If study hours increase, marks may also increase. Linear regression tries to draw the **best-fit straight line** through the data.

\\(\hat{y} = b_0 + b_1x\\)

\\(\hat{y} = \text{8.49} - \text{0.54}x\\)

## 2. Linear Regression Equation

The basic equation is:

\[
\hat{Y} = b_0 + b_1X
\]

Where:

- **Ŷ** = Predicted value of Y
- **X** = Independent variable
- **b₀** = Intercept
- **b₁** = Slope/coefficient

### Example

If:

\[
\hat{Y} = 40 + 5X
\]

and a student studies for **6 hours**:

\[
\hat{Y} = 40 + 5(6)
\]

\[
\hat{Y} = 70
\]

Therefore, the predicted marks are **70**.

---

## 3. Important Terms

### Independent Variable

The variable used to make the prediction.

**Example:** Study hours.

### Dependent Variable

The value that we want to predict.

**Example:** Student marks.

### Slope

The slope tells us how much the predicted Y changes when X increases by one unit.

### Intercept

The predicted value of Y when X is zero.

### Residual/Error

The difference between the actual value and predicted value.

\[
Error = Actual - Predicted
\]

---

## 4. Best-Fit Line

Linear regression tries to find a line that gives the smallest overall prediction error.

The commonly used method is called the **Least Squares Method**.

The idea is to minimize the **sum of squared errors** between the actual points and the values predicted by the line.

## 5. Types of Linear Regression

### A. Simple Linear Regression

Uses **one independent variable**.

Example:

```text
Study Hours → Marks
```

Equation:

\[
Y = b_0 + b_1X
\]

### B. Multiple Linear Regression

Uses **two or more independent variables**.

Example:

```text
Study Hours ──┐
Attendance ───┼──→ Marks
Assignments ──┘
```

Equation:

\[
Y=b_0+b_1X_1+b_2X_2+\cdots+b_nX_n
\]

---

## 6. Steps of Linear Regression

1. Collect the dataset.
2. Identify independent and dependent variables.
3. Split data into training and testing data.
4. Train the linear regression model.
5. Find the best-fit line.
6. Make predictions.
7. Evaluate the model.

---

## 7. Simple Python Example

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Training data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([30, 40, 50, 60, 70])

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict marks for 6 hours
prediction = model.predict([[6]])

print("Predicted Marks:", prediction[0])
```

### Output

```text
Predicted Marks: 80.0
```

Here, the model learns the relationship:

```text
Study Hours → Marks
```

and predicts the marks for a new number of study hours.

## 8. Advantages

- Simple and easy to understand.
- Easy to implement.
- Fast for small datasets.
- Useful for prediction.
- Shows the relationship between variables.

## 9. Disadvantages

- Works best when the relationship is approximately linear.
- Sensitive to outliers.
- Not suitable for every type of data.
- Multiple highly related input variables can cause problems.
- A good fit does not automatically mean a causal relationship.

### Short Exam Definition

**Linear regression is a supervised machine learning algorithm that models the relationship between one or more independent variables and a continuous dependent variable using a linear equation.**

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Prepare sample data (Study Hours vs. Marks)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([30, 42, 48, 62, 70])

# 2. Train the Linear Regression Model
model = LinearRegression()
model.fit(X, y)

# 3. Generate predictions for the regression line
X_line = np.linspace(1, 5, 100).reshape(-1, 1)
y_line = model.predict(X_line)

# 4. Plot the data points and the regression line
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Actual Data Points', s=100)
plt.plot(X_line, y_line, color='red', linewidth=2, label=f'Regression Line (y = {model.coef_[0]:.2f}x + {model.intercept_:.2f})')

# Add titles and labels
plt.title('Visualization of Linear Regression Line', fontsize=14)
plt.xlabel('Study Hours (Independent Variable X)', fontsize=12)
plt.ylabel('Marks (Dependent Variable Y)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.show()
````
